```python
prior_probability_disease = 0.004
n_d = int(prior_probability_disease * 100000)
n_nd = 100000 - n_d
n_pos_d = int(0.99 * n_d)
n_neg_d = n_d - n_pos_d
n_pos_nd = int(0.005 * n_nd)
n_neg_nd = n_nd - n_pos_nd
condition = np.array(['Disease']*n_d + ['No Disease']*n_nd)
test_results_d = np.array(['Positive']*n_pos_d + ['Negative']*n_neg_d)
test_results_nd = np.array(['Positive']*n_pos_nd + ['Negative']*n_neg_nd)
test = np.append(test_results_d, test_results_nd)
t = Table().with_columns(
    'True Condition', condition,
    'Test Result', test
)
t
```

```python
t.pivot('Test Result', 'True Condition')
```
