```python
# Crushed ICE: calculate the total price after applying 8.75% sales tax to the Price column
```

```python
# option 1
tax = cones.column('Price') * (8.75 / 100)
tax + cones.column('Price')
```

```python
# option 2
import numpy as np
np.round(cones.column('Price') * 1.0875, 2)
```

```python
tax_rate = 0.0875
tax = cones.column('Price') * tax_rate
cones_with_tax = cones.with_column('Price with tax', tax + cones.column('Price'))
cones_with_tax
```

```python
cones_with_tax.set_format('Price with tax', CurrencyFormatter)
```

```python
cones_with_tax.column('Price with tax')
```
