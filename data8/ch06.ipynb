```python
pokemon = Table.read_table('data/pokemon.csv')
pokemon
```

```python
temp = pokemon.select('name', 'height_m')
temp
```

```python
temp.where('height_m', are.above_or_equal_to(0)).sort('height_m', descending=True).show(5)
```

```python
temp = pokemon.select('name', 'speed', 'weight_kg').with_column(
    'speed/weight ratio', pokemon.column('speed') / pokemon.column('weight_kg')
)
temp
```

```python
# nan is short for "not a number". It's a value that is a placeholder for "I don't know"
# you can use.where to get rid of nans
temp.where(3, are.above_or_equal_to(0)).sort(3, descending=True)
```

```python
pokemon.where('weight_kg', are.above_or_equal_to(0)).column('weight_kg').mean()
```

```python
# question: how does the speed of pokemon that are more likely to be female
# compare to the speed of pokemon that are more likely to be male?
temp = pokemon.select('name', 'percentage_male', 'speed')
temp
```

```python
temp = temp.with_column('percentage_female', 100 - temp.column('percentage_male'))
temp
```

```python
temp.where('percentage_female', are.above(50))
```

```python
temp.where('percentage_female', are.above(50)).column('speed').mean()
```

```python
temp.where('percentage_male', are.above(50)).column('speed').mean()
```

```python
# question: what are all the electric pokemon from generation 1?
temp = pokemon.select('name', 'type1', 'generation').relabeled('type1', 'type')
temp
```

```python
temp.where('type', are.equal_to('electric')).where('generation', 1)
```

```python
cereal = Table.read_table('data/cereal.csv')
cereal.show(15)
```

```python
shelves = cereal.select('name', 'shelf', 'sugars')
shelves
```

```python
shelf_1 = shelves.where('shelf', are.equal_to(1))
shelf_1
```

```python
shelf_1.column('sugars').mean()
```

```python
shelves.where('shelf', are.equal_to(2)).column('sugars').mean()
```

```python
shelves.where('shelf', are.equal_to(3)).column('sugars').mean()
```

```python
cold_vs_hot = cereal.select('name', 'type', 'protein', 'fiber')
cold_vs_hot
```

```python
# what is the avg. amount of protein in the hot cereals in the dataset?
cold_vs_hot.where('type', are.equal_to('H')).column('protein').mean()
```

```python
cold_vs_hot.where('type', are.equal_to('H')).column('fiber').mean()
```

```python
cold_vs_hot.where('type', are.equal_to('C')).column('protein').mean()
```

```python
cold_vs_hot.where('type', are.equal_to('C')).column('fiber').mean()
```
