```python
def sign(x):
    if x < 0:
        # True branch
        return 'Negative'
    else:
        # False branch
        return 'Non-negative'
```

```python
sign(-42)
```

```python
sign(5)
```

```python
def sign(x):
    if x < 0:
        return 'Negative'
    elif x > 0:
        return 'Positive'
    else: # the only other option is x == 0
        return 'Neither'
```

```python
sign(42)
```

```python
sign(-42)
```

```python
sign(0)
```

# Monty Hall Game

```python
def monty_hall_game():
    '''return a list:
    [original_guess, what door Monty reveals, what door you could switch to]'''

    doors = make_array('goat 1', 'goat 2', 'car')

    guess = np.random.choice(doors)

    if guess == 'goat 1':
        return ['goat 1', 'goat 2', 'car']
    elif guess == 'goat 2':
        return ['goat 2', 'goat 1', 'car']
    else:
        goats = make_array('goat 1', 'goat 2')
        np.random.shuffle(goats)
        return ['car', goats.item(0), goats.item(1)]
```

```python
monty_hall_game()
```

```python
games = Table(make_array('Guess', 'Revealed', 'Remaining')) # Table with column labels but no rows

for i in np.arange(10000):
    games.append(monty_hall_game()) # adds a row to a Table in-place

games
```

```python
original_choice = games.group('Guess')
original_choice
```

```python
remaining_door = games.group('Remaining')
remaining_door
```

```python
joined = original_choice.join('Guess', remaining_door, 'Remaining')
joined = joined.relabeled(0, 'Item').relabeled(1, 'Original door').relabeled(2, 'Door you could switch to')
joined
```

```python
joined.barh(0)
```
