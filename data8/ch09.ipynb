```python
def sign(x):
    if x < 0:
        return 'Negative'
```

```python
def sign(x):
    if x < 0:
        # True branch
        return 'Negative'
    else:
        # False branch
        return 'Non-negative'
```

```python
sign(-42)
```

```python
sign(5)
```

```python
def sign(x):
    if x < 0:
        return 'Negative'
    elif x > 0:
        return 'Positive'
    else: # the only other option is x == 0
        return 'Neither'
```

```python
sign(42)
```

```python
sign(-42)
```

```python
sign(0)
```

```python
grade = 75.5

if grade >= 90:
    print('A')
elif grade >= 80: # if we get here, we know the first condition failed (so grade < 90)
    print('B')
elif grade >= 70:
    print('C')
elif grade >= 60:
    print('D')
else:
    print('F')
```

```python
def flip_100_coins_and_count_heads():
    outcomes = np.random.choice(coin, 100)
    num_heads = np.count_nonzero(outcomes == 'heads')
    return num_heads
```

```python
flip_100_coins_and_count_heads()
```

```python
num_repetitions = 20000

head_counts = make_array()

for i in np.arange(num_repetitions):
    head_count = flip_100_coins_and_count_heads()
    head_counts = np.append(head_counts, head_count)
```

```python
results = Table().with_columns(
    'Repetition', np.arange(1, num_repetitions+1),
    'Number of heads', head_counts
)
results
```
